```python
import os
import pandas as pd
from datetime import datetime

# Define input and output folders
input_folder = "Eye_centers_AI_predicted"
output_folder = "Eye_centers_AI_predicted_transformed"

# Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Dictionary to store DataFrames grouped by the first three digits
master_files = {}

# Process each CSV file in the input folder
for filename in os.listdir(input_folder):
    if filename.endswith(".csv"):
        file_path = os.path.join(input_folder, filename)
        
        # Read the CSV file
        df = pd.read_csv(file_path)
        
        # Extract the first three digits from the filename (assuming filename starts with digits)
        first_three_digits = filename[:3]
        
        # Add the DataFrame to the corresponding group
        if first_three_digits in master_files:
            master_files[first_three_digits].append(df)
        else:
            master_files[first_three_digits] = [df]

# Process each group and create master files
for first_three_digits, dfs in master_files.items():
    # Combine all DataFrames in the group
    combined_df = pd.concat(dfs, ignore_index=True)
    
    # Remove duplicates
    combined_df = combined_df.drop_duplicates()
    
    # Add new columns based on the 'filename' column
    if 'filename' in combined_df.columns:
        # Participant column (first part before /)
        combined_df['Participant'] = combined_df['filename'].str.split('/').str[0]
        
        # Extract date and time parts
        datetime_parts = combined_df['filename'].str.extract(
            r'(\d{4}-\d{2}-\d{2}) (\d{2}:\d{2}:\d{2}\.\d+)')
        
        # Date column (formatted as YYYY-MM-DD)
        combined_df['date'] = datetime_parts[0]
        
        # Timestamp column (formatted as HH/MM/SS.microseconds)
        combined_df['timestamp'] = datetime_parts[1].str.replace(':', '/')
    
    # Save the master file
    output_filename = f"{first_three_digits}.csv"
    output_path = os.path.join(output_folder, output_filename)
    combined_df.to_csv(output_path, index=False)

print("Processing complete! Transformed files saved in:", output_folder)
```

```python
import os
import pandas as pd
from glob import glob

# Configuration
input_roots = [
        "Model_input/Blinks_dataset_flat_test",
        "Model_input/Blinks_dataset_flat_train",
        "Model_input/Open_eyes_dataset_flat_test",
        "Model_input/Open_eyes_dataset_flat_train"
        # Add more folders as needed
    ]
output_folder = "JPEG_filenames"

# Create output folder
os.makedirs(output_folder, exist_ok=True)

# Process each root folder
for root in input_roots:
    # Find all numbered subfolders (001, 002, etc.)
    for subfolder in os.listdir(root):
        subfolder_path = os.path.join(root, subfolder)
        
        # Only process if it's a directory with numeric name
        if os.path.isdir(subfolder_path) and subfolder.isdigit():
            # Find all JPEG files
            jpegs = glob(os.path.join(subfolder_path, '*.jpg')) + \
                    glob(os.path.join(subfolder_path, '*.jpeg'))
            
            # Create DataFrame with just filenames
            df = pd.DataFrame({'jpeg_filename': [os.path.basename(j) for j in jpegs]})
            
            # Save to CSV
            output_path = os.path.join(output_folder, f"{subfolder}.csv")
            df.to_csv(output_path, index=False)
            print(f"Created {output_path} with {len(df)} JPEGs")

print("Done processing all folders")
```

```python
import os
import pandas as pd
import re

# Configuration
input_folder = "JPEG_filenames"
output_folder = "JPEG_filenames_transformed"

# Create output folder
os.makedirs(output_folder, exist_ok=True)

# Process each CSV file
for csv_file in os.listdir(input_folder):
    if not csv_file.endswith('.csv'):
        continue
    
    input_path = os.path.join(input_folder, csv_file)
    output_path = os.path.join(output_folder, csv_file)
    
    # Read the CSV file
    df = pd.read_csv(input_path)
    
    # Function to extract timestamp with exactly 6 decimals
    def extract_timestamp(filename):
        # Match pattern: HH:MM:SS.######
        match = re.search(r'(\d{2}:\d{2}:\d{2}\.\d{6})', filename)
        if match:
            # Convert to format HH/MM/SS.######
            return match.group(1).replace(':', '/')
        return None
    
    # Add timestamp column
    df['timestamp'] = df['jpeg_filename'].apply(extract_timestamp)
    
    # Save the new CSV
    df.to_csv(output_path, index=False)
    print(f"Processed {csv_file} - added timestamp column")

print("All files processed successfully")
```

```python
import os
import pandas as pd

# Configuration
ai_folder = "Eye_centers_AI_predicted_transformed"
jpeg_folder = "JPEG_filenames_transformed"
output_folder = "Eye_centers_AI_predicted_transformed_master_key"

# Create output folder
os.makedirs(output_folder, exist_ok=True)

# Process each matching CSV file
for csv_file in os.listdir(ai_folder):
    if not csv_file.endswith('.csv'):
        continue
    
    # Paths to the files
    ai_path = os.path.join(ai_folder, csv_file)
    jpeg_path = os.path.join(jpeg_folder, csv_file)
    
    # Check if matching JPEG CSV exists
    if not os.path.exists(jpeg_path):
        print(f"No matching JPEG CSV found for {csv_file}")
        continue
    
    print(f"Processing {csv_file}...")
    
    # Read both CSV files
    ai_df = pd.read_csv(ai_path)
    jpeg_df = pd.read_csv(jpeg_path)
    
    # Convert timestamp columns to string type for consistent comparison
    ai_df['timestamp'] = ai_df['timestamp'].astype(str).str.strip()
    jpeg_df['timestamp'] = jpeg_df['timestamp'].astype(str).str.strip()
    
    # Perform inner join on timestamp
    merged_df = pd.merge(
        ai_df,
        jpeg_df,
        on='timestamp',
        how='inner'
    )
    
    # Save the merged result
    output_path = os.path.join(output_folder, csv_file)
    merged_df.to_csv(output_path, index=False)
    print(f"Saved merged file with {len(merged_df)} rows to {output_path}")

print("\nAll matching files processed")
```

```python
import os
import pandas as pd
import random

# Configuration
master_key_folder = "Eye_centers_AI_predicted_transformed_master_key"
image_roots = [
        "Model_input/Blinks_dataset_flat_test",
        "Model_input/Blinks_dataset_flat_train",
        "Model_input/Open_eyes_dataset_flat_test",
        "Model_input/Open_eyes_dataset_flat_train"
        # Add more folders as needed
    ]  # Your original L folders
sample_size = 1  # Number of random CSVs to check

def verify_jpeg_existence():
    # Get all CSV files in master key folder
    csv_files = [f for f in os.listdir(master_key_folder) if f.endswith('.csv')]
    
    if not csv_files:
        print("No CSV files found in master key folder")
        return
    
    # Select random CSV(s) to verify
    selected_files = random.sample(csv_files, min(sample_size, len(csv_files)))
    
    for csv_file in selected_files:
        print(f"\nVerifying {csv_file}...")
        
        # Get the folder ID (001, 003 etc.)
        folder_id = os.path.splitext(csv_file)[0]
        
        # Read the master key CSV
        df = pd.read_csv(os.path.join(master_key_folder, csv_file))
        
        if 'jpeg_filename' not in df.columns:
            print("Error: 'jpeg_filename' column not found")
            continue
        
        # Find all possible locations for these JPEGs
        possible_paths = []
        for root in image_roots:
            subfolder_path = os.path.join(root, folder_id)
            if os.path.exists(subfolder_path):
                possible_paths.append(subfolder_path)
        
        if not possible_paths:
            print(f"No matching {folder_id} folders found in image roots")
            continue
        
        print(f"Checking in folders: {possible_paths}")
        
        # Check each JPEG existence
        total_jpegs = len(df)
        found_jpegs = 0
        
        for jpeg_name in df['jpeg_filename']:
            jpeg_found = False
            
            for folder_path in possible_paths:
                jpeg_path = os.path.join(folder_path, jpeg_name)
                if os.path.exists(jpeg_path):
                    found_jpegs += 1
                    jpeg_found = True
                    break
            
            # Uncomment to see verification details
            # status = "FOUND" if jpeg_found else "MISSING"
            # print(f"{status}: {jpeg_name}")
        
        print("\nVerification Results:")
        print(f"Total entries in CSV: {total_jpegs}")
        print(f"JPEGs physically found: {found_jpegs}")
        print(f"Match rate: {found_jpegs/total_jpegs:.2%}")

if __name__ == "__main__":
    verify_jpeg_existence()
```
